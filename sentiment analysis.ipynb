This project uses the **TF-IDF** method and the **SVM (Support Vector Machine)** algorithm for sentiment analysis of text. The model predicts the overall sentiment by examining the frequency patterns and importance of words within the text.

Another important point is that this model does **not truly understand the meaning of words or sentences**. Instead, it makes predictions based only on the patterns and features it learned during training. As a result, it may perform poorly on complex sentences, sarcasm, irony, and nuanced semantic combinations, so we should not expect too much from it in such cases.

To genuinely understand the meaning and context of words, we need to use **Transformer-based models**, which are specifically designed to capture semantic relationships and contextual information within language.

```python
x=pd.read_csv("sentiment_balanced4.csv")
x=x.sample(frac=0.03,random_state=42)
```

```python
y=x['Score']
y
```

```python
x2=x["Text"]
```

max_features پارامتر  باعث می‌شه فقط تعداد مشخصی از مهم‌ترین کلمات انتخاب بشن و ویژگی ها محدود بشه

The **max_features** parameter limits the number of features by selecting only a specified number of the most important words.


```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x2,y,test_size=0.2,random_state=3,stratify=y)


print('train set:',x_train.shape,y_train.shape)
print('test set:',x_test.shape,y_test.shape)
```

```python
print(x2.shape)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

t = TfidfVectorizer(
    max_features=15000,
    ngram_range=(1,2))

x_train = t.fit_transform(x_train)
x_test = t.transform(x_test)
```

```python
from sklearn.svm import LinearSVC
model=LinearSVC(class_weight='balanced')
model.fit(x_train,y_train)
```

```python
y_perd=model.predict(x_test)
y_perd
```



نمره 1 یعنی مدل دقت بسیار بالایی در پیش بینی داره.

A score of 1 indicates that the model has extremely high accuracy in prediction.



در این قسمت هم عملکرد عالی داشته.

It has also performed excellently in this section.


```python
from sklearn.metrics import  accuracy_score,recall_score,precision_score
accuracy_score(y_test,y_perd)
```

در این قسمت هم عملکرد عالی داشته.

It has also performed excellently in this section.


```python
recall_score(y_test,y_perd,average='micro')
```

در این قسمت هم عملکرد عالی داشته.

It has also performed excellently in this section.


```python
precision_score(y_test,y_perd,average='macro')
```



بطور کلی مدل عملکرد بسیار خوبی داشته است.
