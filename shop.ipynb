```python
x=pd.read_csv('marketing_campaign.csv')
x
```

```python
x.isnull().sum()
```

```python
x=x.dropna()
x
```

```python
x=x.drop('ID',axis=1)

x
```

```python
# x=np.asanyarray(x)
```

```python
from sklearn.preprocessing import LabelEncoder

lb=LabelEncoder()
x['Education']=lb.fit_transform(x['Education'])


lb2=LabelEncoder()
x['Marital_Status']=lb2.fit_transform(x['Marital_Status'])
```

```python
from sklearn.preprocessing import MinMaxScaler
sc=MinMaxScaler()
x=sc.fit_transform(x)
x
```

```python
from sklearn.decomposition import PCA
pca=PCA(n_components=2)
x=pca.fit_transform(x)
x
```

```python
from sklearn.metrics.pairwise import euclidean_distances
dist_matrix = euclidean_distances(x,x)
print(dist_matrix)
```

```python
from sklearn.cluster import AgglomerativeClustering
cluster_n=4
model=AgglomerativeClustering(linkage='average',n_clusters=cluster_n).fit(x)
print(model.labels_)
```

```python
label_ll=model.labels_
label_ll
```



در این قسمت خوشه بندی نمره خوبی دریافت نکرده است وضعیف عمل کرده است.

```python
unique,counts=np.unique(model.labels_, return_counts=True)
print(dict(zip(unique,counts)))
```

```python
df=pd.read_csv('marketing_campaign.csv')
df=df.dropna()
df=df.drop('ID',axis=1)

df['Cluster']=model.labels_

cluster_summary=df.groupby('Cluster').mean(numeric_only=True)
print(cluster_summary.T)
```

```python
total_income=df.groupby(model.labels_)['Income'].sum()

for cluster, income in total_income.items():
    print(f"Cluster {cluster} : {income:,.2f}")
```

```python
df["all_mtn"]=df["MntWines"]+df["MntFruits"]+df["MntMeatProducts"]+df["MntFishProducts"]+df["MntSweetProducts"]+df["MntGoldProds"]
df["child_numbers"]=df["Kidhome"]+df["Teenhome"]

a_mtn=df.groupby("Cluster")['all_mtn'].mean()
a_child_num=df.groupby("Cluster")['child_numbers'].mean()

for i in  a_mtn.index:
    print(f"mean:{i} cluster mtn:{a_mtn[i]},child_mean:{a_child_num[i]}")
```

Group 2: lower income compared to the other groups, and also had the lowest expenditure. They did participate in the survey either.

Group 3: income lower than the other groups, but expenditure is drastically higher. They did not participate in the survey.
